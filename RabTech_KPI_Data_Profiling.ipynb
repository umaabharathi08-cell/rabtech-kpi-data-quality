```python
import pandas as pd
import numpy as np

# 1. Load Raw Dataset
df_orders = pd.read_csv('retail-orders-raw.csv')
print("=== RAW DATASET ===")
print(df_orders)

print("\n" + "="*50)
print("=== DATA QUALITY PROFILING REPORT ===")
print("="*50)

# Check 1: Completeness (Missing Values)
print("\n[1] COMPLETENESS CHECKS:")
missing_summary = df_orders.isnull().sum()
print("Missing values per column:\n", missing_summary[missing_summary > 0])

# Check 2: Uniqueness (Duplicate Records)
print("\n[2] UNIQUENESS CHECKS:")
duplicates = df_orders[df_orders.duplicated(subset=['order_id'], keep=False)]
print(f"Duplicate order_id records found ({len(duplicates)} rows):\n", duplicates[['order_id', 'order_date', 'customer_segment']])

# Check 3: Validity (Value Ranges & Types)
print("\n[3] VALIDITY CHECKS:")
# Quantity check
invalid_quantity = df_orders[pd.to_numeric(df_orders['quantity'], errors='coerce').isna() | (pd.to_numeric(df_orders['quantity'], errors='coerce') <= 0)]
print("Invalid quantity values:\n", invalid_quantity[['order_id', 'quantity']])

# Discount range check (0-100%)
invalid_discount = df_orders[(df_orders['discount_pct'] < 0) | (df_orders['discount_pct'] > 100)]
print("\nInvalid discount percentage (>100% or <0%):\n", invalid_discount[['order_id', 'discount_pct']])

# Date check
df_orders['parsed_date'] = pd.to_datetime(df_orders['order_date'], format='%Y-%m-%d', errors='coerce')
invalid_dates = df_orders[df_orders['order_date'].notna() & df_orders['parsed_date'].isna()]
print("\nInvalid date formats/values:\n", invalid_dates[['order_id', 'order_date']])

# Check 4: Consistency (Categorical Casing)
print("\n[4] CONSISTENCY CHECKS:")
print("Customer Segment Unique Values:", df_orders['customer_segment'].unique())
print("Payment Status Unique Values:", df_orders['payment_status'].unique())
```
